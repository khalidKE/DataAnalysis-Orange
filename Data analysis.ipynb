```python
labels = df["loan_type"].value_counts().index
print(labels)
sizes = df["loan_type"].value_counts()
print(sizes)
colors = ["pink","yellow","blue","green"]
plt.pie(sizes, labels=labels,autopct='%1.1f%%',colors=colors)
plt.title('Loan Types')
plt.show()
```

```python
df["amount_category"] = pd.qcut(df["loan_amount"], q=3, labels=["Low", "Medium", "High"])
df.head()
```

```python
crosstab_data = pd.crosstab(df["repaid"], df["amount_category"])

crosstab_data.plot(kind="bar", figsize=(10, 6), color=["blue", "red", "green"])

plt.title("Repaid Count by Amount Category")
plt.xlabel("Repaid")
plt.ylabel("Count")
plt.legend(title="Amount Category")

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(22,20))
corr1   = df.select_dtypes(include=['float64', 'int64']).corr()
sns.heatmap(corr1, ax=ax, annot=True, cmap='coolwarm', fmt=".2f", linewidths=1, vmin=-1, vmax=1)
plt.xticks(rotation=45, ha='right', fontsize=12)
plt.yticks(fontsize=12)
plt.show()
```

```python
df_repaid = df[df['repaid'] == 1]
df_repaid.head()
```

```python
df_not_repaid = df[df['repaid'] == 0]
df_not_repaid.head()
```

```python
sns.histplot(data=df_repaid['loan_amount'], color='green', label='Repaid')

sns.histplot(data=df_not_repaid['loan_amount'], color='red', label='Not Repaid')

plt.xlabel('Loan Amount')
plt.ylabel('Frequency')
plt.title('Distribution of Loan Amounts by Repayment Status')
plt.legend()
plt.show()
```

```python
sns.histplot(data=df_repaid['rate'], color='green', label='Repaid')

sns.histplot(data=df_not_repaid['rate'], color='red', label='Not Repaid')

plt.xlabel('Interest Rate')
plt.ylabel('Frequency')
plt.title('Distribution of Interest Rates by Repayment Status')
plt.legend()
plt.show()
```
